```python
# 1. SatisfactionIndex – Average of 3 satisfaction metrics
df['SatisfactionIndex'] = (
    df['JobSatisfaction'] + df['EnvironmentSatisfaction'] + df['RelationshipSatisfaction']
) / 3

# 2. TenureBucket – Grouping YearsAtCompany
df['TenureBucket'] = pd.cut(
    df['YearsAtCompany'],
    bins=[0, 2, 5, 10, df['YearsAtCompany'].max()],
    labels=['<2yrs', '2–5yrs', '5–10yrs', '10yrs+']
)

# 3. WorkBalanceScore – Combining WorkLifeBalance & JobInvolvement
df['WorkBalanceScore'] = (
    df['WorkLifeBalance'] + df['JobInvolvement']
) / 2

# 4. IncomeLevel – Binned MonthlyIncome
df['IncomeLevel'] = pd.qcut(df['MonthlyIncome'], q=3, labels=['Low', 'Medium', 'High'])

# 5. Overworked – Flag if experience outside company > 5 years
df['Overworked'] = (df['TotalWorkingYears'] - df['YearsAtCompany']) > 5
df.to_csv('/content/drive/MyDrive/employee_attrition_modified.csv', index=False)


```
